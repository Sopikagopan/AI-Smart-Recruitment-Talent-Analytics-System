#### Univariate Analysis

##### Experience distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df["candidate_experience_years"], bins=13, kde=True)
plt.title("Distribution of Candidate Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Number of Candidates")
plt.show()
```

##### Education distribution

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="candidate_education")
plt.title("Candidate Education Distribution")
plt.xlabel("Education")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

##### Candidate role distribution

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, y="candidate_role")
plt.title("Candidate Role Distribution")
plt.xlabel("Count")
plt.ylabel("Candidate Role")
plt.show()
```

##### Job title distribution

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, y="job_title")
plt.title("Job Title Distribution")
plt.xlabel("Count")
plt.ylabel("Job Title")
plt.show()
```

##### Target distribution

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="match_label")
plt.title("Candidate-Job Match Distribution")
plt.xlabel("Match Label")
plt.ylabel("Count")
plt.show()
```

#### Bivariate Analysis

##### Experience vs Match

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="match_label", y="candidate_experience_years")
plt.title("Candidate Experience vs Job Match")
plt.xlabel("Match Label")
plt.ylabel("Candidate Experience (Years)")
plt.show()
```

##### Education vs Match

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="candidate_education", hue="match_label")
plt.title("Education Level vs Job Match")
plt.xlabel("Education")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

##### Candidate Role vs Match

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df, y="candidate_role", hue="match_label")
plt.title("Candidate Role vs Job Match")
plt.xlabel("Count")
plt.ylabel("Candidate Role")
plt.show()
```

##### Job Title vs Match

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df, y="job_title", hue="match_label")
plt.title("Job Title vs Job Match")
plt.xlabel("Count")
plt.ylabel("Job Title")
plt.show()
```

#### Correlation Analysis

###### For the numerical variables:


```python
correlation = df[numerical_cols].corr()

plt.figure(figsize=(7,5))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

#### Outlier Analysis

##### For the numerical variables:

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="candidate_experience_years")
plt.title("Candidate Experience Outlier Analysis")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="job_required_experience")
plt.title("Required Experience Outlier Analysis")
plt.show()
```
